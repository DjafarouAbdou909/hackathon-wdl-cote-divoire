### Objectif de ce notebook
Explorer et comprendre les trois sources de données disponibles
avant toute transformation ou analyse.

**Datasets utilisés :**
- `raw_africa_clock_15_35.xlsx` — Africa Youth Employment Clock (World Data Lab)
- `Data_africa_education_unemployed.csv` — Chômage par éducation (ILO)
- `Data_africa_sector_employed.csv` — Emploi par secteur (ILO)

```python
# IMPORTS

import pandas as pd
import os
```

```python
# PATHS

path_raw = os.path.join('..', 'data', 'raw')

file_clock = os.path.join(path_raw, 'raw_africa_clock_15_35.xlsx')
file_edu_unemp = os.path.join(path_raw, 'Data_africa_education_unemployed.csv')
file_sector = os.path.join(path_raw, 'Data_africa_sector_employed.csv')
```

```python
# LOAD DATASETS

df_clock = pd.read_excel(file_clock)

df_edu_unemp = pd.read_csv(file_edu_unemp)

df_sector = pd.read_csv(file_sector)
```

```python
# DATASET

datasets = {
    "clock": df_clock,
    "education_unemployment": df_edu_unemp,
    "sector_employment": df_sector
}
```

### Vue d'ensemble des datasets
Vérification des colonnes, types et premières lignes
pour chaque source de données.

```python
# OVERVIEW OF DATASETS

for name, df in datasets.items():

    print(f"{'='*60}")
    print(f"DATASET : {name.upper()}")
    print(f"{'='*60}")

    print("Columns:")
    print(df.columns.tolist())

    print("First 3 rows:")
    display(df.head(3))
```

```python
# DATASET INFO

for name, df in datasets.items():

    print(f"{'='*60}")
    print(f"INFO : {name.upper()}")
    print(f"{'='*60}")

    df.info()
```

```python
# 7. DESCRIPTION

for name, df in datasets.items():

    print(f"{'='*60}")
    print(f"STATISTICS : {name.upper()}")
    print(f"{'='*60}")

    display(df.describe())
```

### Qualité des données
Vérification des valeurs manquantes et des doublons.

> Résultat attendu : zéro valeur manquante, zéro doublon 
> les données sont prêtes pour la préparation.

```python
# MISSING VALUES

for name, df in datasets.items():

    print(f"{name.upper()} - Missing values")

    print(df.isnull().sum())
```

```python
# DUPLICATES

for name, df in datasets.items():

    duplicates = df.duplicated().sum()

    print(f"\n{name.upper()} - Duplicates: {duplicates}")
```

###  Conclusion
Les trois datasets sont complets et sans anomalie.

éducation/secteur, et **2026** pour le snapshot Africa Clock.
