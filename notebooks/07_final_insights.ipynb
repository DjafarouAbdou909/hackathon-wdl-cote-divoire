# FINAL INSIGHT — Sans transformation structurelle accélérée, le dividende démographique ivoirien risque de se convertir en vulnérabilité de masse d'ici 2035

> *Note méthodologique préliminaire : Les projections présentées reposent sur un modèle de décroissance exponentielle appliqué à un indice composite de vulnérabilité (chômage + inactivité des jeunes), calibré sur des données historiques 2015–2025. Les taux de décroissance retenus par scénario (0,5 %, 1,5 % et 3 %) constituent des hypothèses paramétriques, non des prévisions économétriques structurelles. Leur interprétation doit donc demeurer prospective et conditionnelle, non déterministe.*

## 1. Résultats clés des projections : une divergence de trajectoires lourde de sens

Les trois scénarios simulés à l'horizon 2035 révèlent une réalité que les chiffres agrégés de croissance du PIB ivoirien tendent à masquer : **l'ampleur de la réduction de la vulnérabilité juvénile dépend de façon non linéaire de l'intensité des politiques publiques engagées**.

À partir d'un indice de vulnérabilité initial de 0,480 en 2025, les projections montrent des trajectoires radicalement différenciées :

- **Scénario 1 — Inertie structurelle** (taux de décroissance : 0,5 %/an) : L'indice atteint 0,457 en 2035, soit une réduction de seulement **4,9 %** sur une décennie. Cette trajectoire correspond à la continuation de la dynamique observée depuis 2020, caractérisée par une quasi-stagnation de l'indice autour de 0,48. En termes absolus, une réduction de moins de 5 % de la vulnérabilité sur dix ans signifie que près d'un jeune sur deux resterait en situation de chômage ou d'inactivité en 2035 — un résultat économiquement et socialement intenable dans un pays où la population en âge de travailler s'accroît à un rythme soutenu.

- **Scénario 2 — Réformes partielles** (taux de décroissance : 1,5 %/an) : L'indice descend à 0,413, soit une réduction de **14 %**. Ce scénario représente un progrès réel mais insuffisant. Il correspond à la mise en œuvre de politiques sectorielles d'amélioration — formation professionnelle, soutien à l'entrepreneuriat, extension partielle de la protection sociale — sans remise en cause des structures profondes du marché du travail. La vulnérabilité diminue, mais à un rythme qui ne compense pas l'arrivée massive de nouvelles cohortes sur le marché de l'emploi.

- **Scénario 3 — Transformation structurelle** (taux de décroissance : 3 %/an) : L'indice atteint 0,354 en 2035, soit une réduction de **26,3 %**. C'est le seul scénario qui modifie véritablement la logique de reproduction de la vulnérabilité. L'écart avec le scénario d'inertie représente, en valeur absolue d'indice, plus de 0,10 point — un écart considérable lorsqu'il est rapporté aux millions de jeunes concernés.

L'écart entre le scénario 1 et le scénario 3 — de l'ordre de 21 points de pourcentage de réduction cumulée — constitue **la mesure du coût d'opportunité de l'inaction**. Ce différentiel n'est pas marginal : il représente la différence entre une économie capable de valoriser son capital humain jeune et une économie qui le neutralise structurellement.

## 2. Interprétation structurelle : pourquoi la croissance seule ne suffit pas

La quasi-stagnation de l'indice de vulnérabilité sur la période historique 2015–2025, malgré des taux de croissance du PIB soutenus en Côte d'Ivoire, illustre un phénomène bien documenté dans la littérature sur le développement africain : **la déconnexion entre croissance agrégée et création d'emplois formels de qualité**.

Cette déconnexion s'explique par plusieurs mécanismes structurels imbriqués :

**L'informalité comme structure dominante et reproductrice.** L'économie ivoirienne repose très largement sur un secteur informel qui absorbe les flux de nouveaux entrants sur le marché du travail sans pour autant constituer un vecteur de productivité ou de protection sociale. L'informalité n'est pas uniquement un état transitoire vers l'emploi formel ; elle est, dans ce contexte, une configuration d'équilibre stable, alimentée par l'insuffisance du secteur formel à générer des emplois en volume et en qualité suffisants. Toute politique qui ne s'attaque pas aux déterminants structurels de l'informalité — fiscalité, coût du travail, accès au financement des PME, régulation du marché du travail — restera cantonnée à des effets marginaux.

**L'absence de diversification productive.** La structure sectorielle de l'économie ivoirienne reste très concentrée autour de l'agriculture (notamment le cacao) et des services informels. L'industrie manufacturière, vecteur historique de création d'emplois formels à grande échelle dans les transitions réussies (Asie du Sud-Est, Maroc, Éthiopie partiellement), demeure structurellement sous-développée. Sans montée en gamme productive — transformation locale des matières premières, développement d'industries légères employant intensivement la main-d'œuvre — la demande de travail formel restera insuffisante face à l'offre croissante de jeunes diplômés ou qualifiés.

**Le déficit d'adéquation formation-emploi.** L'élévation du niveau d'instruction des jeunes ivoiriens, réelle sur la période récente, ne s'est pas accompagnée d'une transformation équivalente de la structure de la demande de compétences. La conséquence est une double pression : déclassement des diplômés dans des emplois informels sous-qualifiés, et persistance de pénuries de compétences techniques dans des secteurs porteurs. Ce désajustement structurel nourrit à la fois le découragement des jeunes et la sous-utilisation du capital humain accumulé.

**La sous-inclusion des femmes dans l'emploi formel.** L'indice de vulnérabilité composite intègre l'inactivité, qui affecte de manière disproportionnée les femmes. Leur faible participation au marché du travail formel n'est pas simplement une question de préférences individuelles : elle reflète des contraintes structurelles (charge de travail non rémunéré, discrimination à l'embauche, déficit de structures d'accueil de la petite enfance, normes sociales) qui ne peuvent être levées que par des politiques publiques délibérées et multi-dimensionnelles. Ignorer cette dimension revient à renoncer à mobiliser une fraction déterminante du potentiel productif national.

## 3. Risques macroéconomiques et sociaux : les conséquences de l'inertie

Si la trajectoire d'inertie structurelle devait prévaloir, les implications s'étendraient bien au-delà du marché du travail, affectant la soutenabilité même du modèle de développement ivoirien.

**Le dividende démographique comme risque plutôt qu'opportunité.** La Côte d'Ivoire bénéficie actuellement d'une fenêtre démographique favorable, avec une proportion croissante de population en âge de travailler. Cette configuration n'est productive que si le système économique est en mesure d'absorber et de valoriser cette main-d'œuvre. En l'absence de transformation structurelle, la même dynamique démographique qui constitue un dividende potentiel devient un facteur d'amplification des tensions : afflux massif de jeunes sur un marché du travail incapable de les intégrer, pression sur les systèmes de protection sociale, montée du sous-emploi visible et invisible.

**Risques de stabilité sociale.** Une vulnérabilité persistante à hauteur de 45–48 % de la population jeune en 2035 (scénario d'inertie) constituerait un facteur de déstabilisation sociale significatif. La littérature empirique sur les conflits et la gouvernance en Afrique subsaharienne établit une corrélation robuste entre chômage et inactivité juvéniles, d'une part, et fragilité institutionnelle, recrutement par des acteurs non étatiques et tensions intercommunautaires, d'autre part. Ce risque est particulièrement aigu dans le contexte ivoirien, où la mémoire des crises politiques passées reste prégnante.

**Pression sur les finances publiques.** Un stock élevé de jeunes inactifs ou en emploi informel génère des externalités fiscales négatives de plusieurs ordres : faible assiette de cotisations sociales, demande accrue de services publics (santé, sécurité), pression à la hausse sur les transferts sociaux, et réduction de la base fiscale formelle. Cette configuration compromet la capacité de l'État à financer les investissements structurants (infrastructure, éducation, santé) qui sont précisément les leviers d'une sortie du scénario d'inertie.

**Érosion de la confiance dans le système éducatif.** Lorsque des cohortes successives de jeunes diplômés constatent que leur investissement éducatif ne se traduit pas par une insertion professionnelle satisfaisante, cela génère un phénomène de dépréciation perçue du capital humain. Cette déconnexion peut conduire à des comportements de sous-investissement éducatif des ménages, notamment pour les filles, et à une remise en cause de la légitimité du système de formation — avec des effets négatifs de long terme sur la productivité agrégée.

**Soutenabilité du modèle de croissance.** La Côte d'Ivoire a enregistré des performances macroéconomiques remarquables depuis 2012. Toutefois, ce modèle de croissance repose en partie sur des fondamentaux conjoncturels (prix des matières premières, investissements publics en infrastructure) dont la soutenabilité est incertaine à l'horizon 2035. Une transition vers une croissance tirée par la productivité et la valeur ajoutée intérieure requiert précisément la mobilisation efficace du capital humain jeune — une ressource que le scénario d'inertie neutralise systématiquement.

## 4. Lecture comparative des scénarios : trois logiques économiques, trois futurs possibles

La différence entre les trois scénarios ne tient pas uniquement à l'intensité des interventions, mais à leur **nature et à leur cohérence systémique**.

**Pourquoi le scénario 1 échoue structurellement.** Une décroissance annuelle de 0,5 % de la vulnérabilité correspond, dans la pratique, à la continuation des politiques actuelles sans inflexion significative. Ce taux de réduction ne compense pas la pression démographique, le différentiel de productivité entre secteurs formel et informel, ni les inégalités d'accès à l'emploi de qualité. Le scénario 1 n'est pas un scénario de stagnation intentionnelle ; il est la résultante probable d'une accumulation de réformes incrémentales non coordonnées, de priorités budgétaires fragmentées, et d'un déficit de vision transformatrice à long terme. Sa logique économique sous-jacente est celle d'une économie qui croît sans se transformer — ce que la littérature du développement qualifie de « trappe à informalité ».

**Pourquoi le scénario 2 améliore partiellement la situation.** Un taux de décroissance de 1,5 % reflète une politique publique plus active : amélioration des dispositifs de formation professionnelle, programmes d'appui à l'entrepreneuriat, renforcement de l'inspection du travail, mesures incitatives à la formalisation des petites entreprises. Ces interventions produisent des effets réels, documentés dans plusieurs pays africains comparables. Toutefois, leur impact reste limité par un plafond structurel : elles agissent sur les conditions d'accès à l'emploi existant sans modifier fondamentalement la structure de la demande de travail. Le scénario 2 réduit la vulnérabilité à la marge sans en supprimer les causes profondes. C'est, pour employer une métaphore analytique courante dans les policy briefs de l'OCDE, une politique de « redistribution de la pénurie » plutôt que de création de l'abondance.

**Pourquoi le scénario 3 change réellement la trajectoire.** Un taux de décroissance de 3 % suppose des transformations simultanées sur plusieurs fronts — industrialisation accélérée, formalisation massive, inclusion économique des femmes, refonte de l'adéquation formation-emploi, développement d'écosystèmes productifs régionaux. La logique économique ici est celle d'une transformation structurelle au sens de McMillan et Rodrik : un déplacement net de la main-d'œuvre des secteurs à faible productivité vers des secteurs à haute productivité, accompagné d'une hausse de la productivité intra-sectorielle. Ce scénario ne suppose pas une rupture magique ; il requiert une cohérence et une persistance de l'action publique sur une décennie, soutenue par des financements adéquats et une gouvernance de qualité. Sa faisabilité technique est établie par les expériences comparatives ; sa faisabilité politique constitue le véritable défi.

## 5. Conclusion politique majeure : pour un agenda de transformation structurelle de l'emploi des jeunes en Côte d'Ivoire

Les projections analysées conduisent à une conclusion politique qui s'impose avec une netteté analytique rare : **le problème de la vulnérabilité juvénile en Côte d'Ivoire n'est ni individuel, ni principalement éducatif, ni réductible à des déficits de compétences. Il est structurel, systémique, et requiert une réponse de même nature.**

Les politiques qui se concentrent exclusivement sur l'offre de travail — amélioration de la formation, accompagnement à l'insertion, soutien à l'entrepreneuriat individuel — sont nécessaires mais insuffisantes. Elles s'inscrivent dans la logique du scénario 2 : utiles, mesurables, mais incapables de modifier la trajectoire de fond. L'expérience comparée des pays ayant réussi des transitions d'emploi à grande échelle indique qu'une réduction durable et substantielle de la vulnérabilité juvénile requiert une combinaison cohérente de politiques agissant simultanément sur plusieurs leviers :

**Premièrement, l'industrialisation et la diversification productive.** Il s'agit de créer les conditions d'une demande de travail formel suffisante pour absorber les flux d'entrants. Cela passe par des politiques industrielles délibérées, des zones économiques spéciales, des stratégies de chaînes de valeur régionales, et un soutien ciblé aux secteurs à fort potentiel d'emploi formel (agro-industrie, BTP structuré, économie numérique, secteur des services à fort contenu technologique).

**Deuxièmement, la formalisation comme priorité transversale.** La réduction de la vulnérabilité passe nécessairement par l'extension du secteur formel — non pas par l'imposition administrative, mais par la réduction du coût relatif de la formalité pour les entreprises et les travailleurs. Cela implique des réformes de la fiscalité des PME, de la régulation du marché du travail, et de l'extension des systèmes de protection sociale.

**Troisièmement, l'inclusion économique des femmes comme levier de productivité nationale.** Les politiques d'emploi qui ne ciblent pas explicitement les obstacles à la participation féminine au marché du travail formel — congé parental, infrastructures d'accueil de la petite enfance, lutte contre les discriminations à l'embauche, accès au crédit — renoncent à mobiliser une fraction déterminante du capital humain disponible.

**Quatrièmement, la refonte de l'adéquation formation-emploi.** L'expansion quantitative de l'éducation doit être accompagnée d'une réorientation qualitative vers les compétences techniques, numériques et entrepreneuriales que demandent les secteurs porteurs. Cela requiert des mécanismes institutionnalisés de dialogue entre système éducatif et secteur privé, ainsi qu'un investissement soutenu dans la formation professionnelle et l'apprentissage dual.

**Cinquièmement, la cohérence et la persistance de l'action publique.** L'une des leçons les plus robustes de l'économie du développement est que les transformations structurelles ne se produisent pas dans l'espace d'un programme politique ou d'un mandat électoral. Elles requièrent des engagements institutionnels de long terme, des mécanismes de financement stables, et des dispositifs d'évaluation permettant d'ajuster les politiques sans les interrompre.

En définitive, les projections à l'horizon 2035 posent une question qui dépasse la technique économétrique : la Côte d'Ivoire choisira-t-elle de gérer sa vulnérabilité juvénile par des ajustements à la marge, ou de la transformer par une ambition structurelle à la hauteur de l'enjeu démographique ? Les données disponibles indiquent clairement que seule la seconde voie offre une perspective de réduction significative et soutenable de la vulnérabilité. Le coût de l'inaction, mesuré en points d'indice comme en millions de trajectoires de vie compromises, constitue l'argument économique le plus convaincant en faveur d'un agenda de transformation structurelle résolument assumé.
