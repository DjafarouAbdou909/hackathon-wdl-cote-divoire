# INSIGHT 1 — Le Paradoxe de l'Éducation Tertiaire : Un Marché Formel Structurellement Incapable d'Absorber Ses Propres Diplômés

## 1. Constats observés

- Sur la période 2015–2025, l'indice de vulnérabilité moyen (chômage + inactivité) atteint son niveau maximal au palier tertiaire (0.625), suivi du secondaire supérieur (0.599), du secondaire inférieur (0.511), du primaire (0.445) et du sous-primaire (0.310).
- La relation éducation–vulnérabilité est **monotone croissante à partir du niveau primaire**, contredisant la fonction de rendement décroissant du risque attendue par la théorie standard du capital humain.
- Le taux de chômage au niveau tertiaire (0.280 en 2025) est environ dix fois supérieur à celui du niveau sous-primaire (0.018), tandis que les taux d'inactivité restent élevés à tous les niveaux.
- Le mismatch index (notebook 05) quantifie ce paradoxe : le secteur formel ne représente que **5.6%** de l'emploi jeune total, contre 53.5% pour le secteur informel, soit un écart structurel de **47.9 points de pourcentage**.

## 2. Interprétation structurelle

**Hypothèse 1 — Étroitesse du secteur formel comme contrainte d'absorption principale :**  
Le résultat du notebook 05 est décisif : avec seulement 5.6% de l'emploi dans les secteurs formels (éducation, administration, finance, santé, professions libérales), le marché ivoirien ne dispose pas de la capacité structurelle d'absorber les flux de diplômés du supérieur. Ce n'est pas la qualité des formations qui est en cause en première analyse — c'est l'étroitesse de la demande formelle de qualifications. Le chômage des diplômés est ainsi une conséquence mécanique d'un déséquilibre offre-demande au niveau macrostructurel, et non nécessairement un signal d'inadéquation des contenus.

**Hypothèse 2 — Effet de réservation salariale et file d'attente vers l'emploi formel :**  
Les diplômés du supérieur maintiennent des salaires de réservation incompatibles avec l'emploi informel dominant. Cette stratégie rationnelle allonge la durée de recherche et gonfle les taux de chômage déclarés au niveau tertiaire. Ce mécanisme de "file d'attente" (queuing model, Fields, 2011) est cohérent avec la structure observée : taux de chômage tertiaire élevé, taux d'inactivité modéré relatif.

**Hypothèse 3 — Inadéquation qualitative secondaire :**  
Sans désagrégation par filière, cette hypothèse reste non testable dans le cadre présent. Elle ne peut être ni confirmée ni exclue. Des filières universitaires produisant massivement des généralistes dans des disciplines peu demandées par le privé formel (droit, lettres, administration générale) amplifieraient le mismatch quantitatif par un mismatch qualitatif surajouté.

## 3. Limites analytiques

- **Construction du mismatch index :** La classification formel/informel/mixte repose sur une règle textuelle (`str.contains`) appliquée aux intitulés sectoriels ISIC-21. Cette classification est une approximation : elle ne s'appuie pas sur des données d'enregistrement des entreprises, de contrats de travail ou de couverture sociale. La part du "formel" (5.6%) peut être sous-estimée si des emplois formels existent dans des secteurs classés "mixte" ou "informel".
- **Absence de désagrégation par filière universitaire :** Le niveau "tertiaire" agrège toutes les formations post-baccalauréat. L'hétérogénéité interne (BTS technique vs licence générale vs ingénieur) est invisible dans les données disponibles.
- **Non-comparabilité des deux datasets :** Le dataset éducation (330 obs.) et le dataset sectoriel (9 702 obs.) ne partagent pas la même unité d'observation. La mise en relation directe (vulnérabilité tertiaire + mismatch sectoriel) constitue une triangulation analytique, non une preuve statistique formelle.
- **Biais de sélection au niveau tertiaire :** Les jeunes atteignant le supérieur sont socialement sélectionnés (urbains, milieux favorisés). Leur taux de chômage peut partiellement refléter des caractéristiques non observées du contexte.

## 4. Importance systémique

- **Perte de rendement social de l'éducation :** Si le chômage des diplômés du supérieur est structurel et durable, le rendement social de l'investissement public en éducation devient négatif. Chaque cohorte diplômée non absorbée représente un coût budgétaire sans contrepartie productive.
- **Désincitation à l'investissement éducatif privé :** À moyen terme, la visibilité du chômage des diplômés réduit les incitations des ménages à financer les études supérieures de leurs enfants, freinant l'accumulation de capital humain.
- **Contrainte sur la croissance à long terme :** La transformation structurelle vers une économie à plus haute valeur ajoutée requiert une main-d'œuvre qualifiée. Un marché formel trop étroit bloque cette transition en maintenant les diplômés dans l'inactivité ou l'informel, sous-utilisant le capital humain accumulé.
- **Stabilité sociale :** L'écart entre aspirations légitimes portées par le diplôme et réalités du marché du travail constitue un facteur documenté de frustration et de tensions, particulièrement dans les contextes urbains.

## 5. Implications politiques

**Court terme (0–2 ans) — priorité haute :**
- Mettre en place un système d'information public sur les taux d'insertion par filière et établissement à 12 et 24 mois post-diplôme, afin d'orienter les choix d'inscription et de cibler les réformes.
- Déployer des programmes de stage obligatoires co-financés État–entreprises pour les étudiants du supérieur, réduisant les coûts de transaction à l'embauche pour les employeurs formels.

**Court terme (0–2 ans) — priorité moyenne :**
- Créer des subventions temporaires à l'embauche de jeunes diplômés dans les PME formelles, conditionnées à une durée minimale de contrat (≥12 mois).

**Moyen terme (2–5 ans) — priorité haute :**
- Définir une stratégie nationale explicite d'élargissement du secteur formel, avec des objectifs quantifiés : passer de 5.6% à 10–12% de l'emploi jeune dans le formel à horizon 2030, via l'agro-industrie, le numérique et les services modernes.
- Réorienter les capacités d'accueil universitaires vers les filières à fort potentiel d'insertion formelle (ingénierie, numérique, agro-industrie, santé), via des mécanismes d'incitation financière différenciés par filière.

**Moyen terme (2–5 ans) — priorité moyenne :**
- Développer des dispositifs d'alternance liant universités et secteur privé formel, sur des modèles adaptés au contexte ouest-africain (expériences marocaine et sénégalaise documentées).

# INSIGHT 2 — L'Exclusion Féminine Systémique : Une Inactivité Transversale qui Résiste au Capital Humain

## 1. Constats observés

- À chaque niveau d'éducation, le taux d'inactivité féminin est systématiquement supérieur au taux masculin, sans exception observable dans les données 2025.
- L'écart genré en termes d'inactivité est le plus marqué au secondaire supérieur (femmes ~0.65 vs hommes ~0.31) et persiste au niveau tertiaire (femmes ~0.44 vs hommes ~0.33).
- La vulnérabilité globale féminine dépasse systématiquement la vulnérabilité masculine à tous les niveaux d'éducation, l'écart étant principalement porté par le différentiel d'inactivité plutôt que par le différentiel de chômage.
- La ségrégation sectorielle est documentée : l'agriculture et la construction concentrent massivement les hommes ; l'hébergement-restauration et les activités de ménage présentent une composition plus équilibrée ou féminine.

## 2. Interprétation structurelle

**Hypothèse 1 — Normes sociales et contraintes reproductives indépendantes du niveau d'éducation :**  
La persistance de l'écart d'inactivité au niveau tertiaire — là où les femmes ont pourtant investi davantage en capital humain — signale que les normes sociales régissant la participation féminine au marché du travail opèrent avec une force suffisante pour neutraliser l'effet attendu de l'éducation sur la participation. Les contraintes liées à la maternité, aux soins familiaux et aux représentations du rôle de la femme active semblent structurellement déterminantes, indépendamment du niveau de qualification.

**Hypothèse 2 — Ségrégation sectorielle comme barrière structurelle à l'entrée :**  
Les secteurs dominants de l'emploi jeune (agriculture intensive, construction, transport) présentent des conditions — mobilité géographique élevée, pénibilité, horaires atypiques, contextes de sécurité précaires — qui créent des barrières à l'entrée de facto pour les femmes dans le contexte ivoirien. L'inactivité peut alors résulter non d'un refus de travailler mais d'une absence d'offres d'emploi compatibles avec les contraintes féminines réelles.

**Hypothèse 3 — Découragement et retrait invisible du marché :**  
Une fraction de l'inactivité féminine peut refléter un découragement actif post-expériences d'échec d'insertion ou de discrimination. Ce phénomène est structurellement invisible dans les indicateurs de chômage officiel (définition BIT), conduisant à une sous-estimation systématique de l'exclusion féminine effective.

## 3. Limites analytiques

- **Limite critique de la définition BIT :** La variable `share_inact` classe comme "inactives" des femmes exerçant potentiellement des activités économiques non marchandes substantielles (agriculture vivrière, travail domestique, soins). Cette convention comptable internationale sous-estime la contribution économique réelle des femmes et sur-estime leur exclusion productive au sens strict.
- **Absence de désagrégation par tranche d'âge dans l'analyse de genre :** Le dataset `ci_clock_snapshot_2026.csv` (882 obs., tranches 15–24 / 25–35) n'est pas mobilisé dans les notebooks 03 et 05. La distinction entre inactivité en cours d'études (15–24 ans) et inactivité post-insertion (25–35 ans) reste non opérée, masquant des dynamiques d'âge potentiellement importantes.
- **Absence de données sur les raisons de non-participation :** Sans module qualitatif, inactivité contrainte, inactivité choisie et découragement sont indistinguables, rendant toute recommandation politique non ciblée.
- **Hétérogénéité géographique non capturée :** Les dynamiques d'inactivité féminine à Abidjan et dans les zones rurales du Nord sont très probablement différentes. Les données nationales agrégées masquent ces disparités.

## 4. Importance systémique

- **Coût macroéconomique quantifiable :** Les estimations FMI/Banque mondiale pour l'Afrique subsaharienne suggèrent qu'une réduction significative des écarts de participation femmes-hommes pourrait représenter des gains de PIB de 5 à 10% à moyen terme. L'inactivité féminine structurelle constitue ainsi une contrainte macroéconomique, pas seulement une question d'équité.
- **Gaspillage d'investissement public :** L'inactivité des diplômées du supérieur représente un coût budgétaire direct : l'État finance une éducation dont le rendement social est neutralisé par l'exclusion du marché du travail.
- **Transmission intergénérationnelle des inégalités :** Le retrait des femmes du marché du travail formel réduit leur autonomie économique, affecte les trajectoires éducatives de leurs enfants et perpétue des cycles de vulnérabilité documentés dans la littérature sur le développement.
- **Dividende démographique conditionnel :** La fenêtre démographique favorable de la Côte d'Ivoire ne se convertira en accélération de croissance que si les taux de participation féminine augmentent substantiellement — condition actuellement non remplie selon les données.

## 5. Implications politiques

**Court terme (0–2 ans) — priorité haute :**
- Étendre la couverture des services de garde d'enfants subventionnés dans les zones urbaines denses, afin de réduire les contraintes de conciliation travail-famille. Ce levier est le plus directement lié à la contrainte reproductrice identifiée.
- Intégrer des objectifs chiffrés de participation féminine dans tous les programmes publics d'emploi et les marchés publics.

**Court terme (0–2 ans) — priorité moyenne :**
- Conduire une enquête qualitative nationale sur les raisons de non-participation des femmes au marché du travail, désagrégée par tranche d'âge et zone géographique. Sans cette donnée, les politiques resteront non ciblées.

**Moyen terme (2–5 ans) — priorité haute :**
- Réformer les dispositifs de protection maternité pour redistribuer les coûts entre employeurs et État, réduisant les incitations à la discrimination à l'embauche des femmes.
- Développer des programmes de formation professionnelle conçus spécifiquement pour les femmes dans les secteurs en croissance compatibles avec les contraintes identifiées (agro-transformation, numérique à distance, santé communautaire).

**Moyen terme (2–5 ans) — priorité moyenne :**
- Investir dans les infrastructures réduisant la charge de travail domestique non rémunéré (eau potable, énergie fiable, transport sûr), leviers indirects mais documentés de la participation féminine.

# INSIGHT 3 — La Concentration Informelle à 68% : Signal d'une Transformation Structurelle Bloquée

## 1. Constats observés

- Agriculture, forêts et pêche concentrent **48.6%** de l'emploi jeune total (population cumulée : ~28.9 millions).
- Le commerce de gros et de détail représente un second pôle à **18.9%** (~11.2 millions).
- Ces deux secteurs absorbent conjointement **67.5%** de l'emploi jeune ivoirien.
- La fabrication (manufacturing) constitue le troisième secteur à **8.1%**, suivi de l'hébergement-restauration (4.5%), des autres services (4.0%) et du transport (3.4%).
- Éducation, administration publique et finance — secteurs typiquement formels — représentent respectivement 1.5%, 1.4% et moins de 1% de l'emploi jeune total.
- Le secteur informel au sens large (agriculture + construction + ménages) totalise **53.5%** de l'emploi ; le secteur formel strict (éducation + administration + finance + santé + professions libérales) atteint seulement **5.6%**.

## 2. Interprétation structurelle

**Hypothèse 1 — Persistance de la spécialisation agro-exportatrice et déficit de transformation structurelle :**  
La structure sectorielle observée est cohérente avec une économie dont le modèle de croissance reste centré sur les cultures de rente (cacao, cajou, caoutchouc) et le commerce de distribution. En l'absence d'investissements suffisants dans la transformation industrielle des matières premières agricoles, le surplus de main-d'œuvre jeune est absorbé dans l'agriculture de subsistance et le petit commerce informel, à productivité marginale faible. Ce blocage de la transformation structurelle (au sens de McMillan & Rodrik, 2011) maintient l'économie dans un équilibre de sous-emploi productif.

**Hypothèse 2 — Fonction d'absorption de dernier recours de l'informel commercial :**  
La part élevée du commerce de gros et de détail (18.9%) reflète probablement la fonction d'absorption de l'économie informelle urbaine : des jeunes ne trouvant pas d'emploi formel s'insèrent dans le petit commerce comme stratégie de survie plutôt que comme choix optimal. Ce mécanisme gonfle les effectifs sectoriels sans augmentation correspondante de la productivité ou des revenus.

**Hypothèse 3 — Barrières à l'investissement industriel :**  
L'étroitesse du secteur manufacturing (8.1%) dans un pays à forte base agricole suggère des contraintes à l'industrialisation : coûts de l'énergie, logistique insuffisante, accès limité au crédit pour les PME industrielles, ou cadre réglementaire peu incitatif. Sans levée de ces contraintes, la transformation structurelle restera bloquée indépendamment des politiques éducatives.

## 3. Limites analytiques

- **Qualité de l'emploi non mesurée :** Les données sectorielles (`population` par secteur) ne capturent pas la productivité, la rémunération, le statut formel/informel intra-sectoriel, ni la durée du travail. L'agriculture industrielle (plantation Olam) et l'agriculture vivrière de subsistance sont comptabilisées identiquement.
- **Classification formel/informel par règle textuelle :** Le mismatch index repose sur une classification par mots-clés, non validée par des données d'enregistrement ou de couverture sociale. La frontière formel/mixte est particulièrement incertaine pour le manufacturing et le transport.
- **Agrégation sectorielle ISIC-21 trop grossière :** La catégorie agriculture regroupe des réalités économiques extrêmement hétérogènes. Sans sous-catégorisation, les recommandations sectorielles restent nécessairement génériques.
- **Absence de données sur les trajectoires :** Les données transversales agrégées ne permettent pas de distinguer jeunes en emploi de passage (début de trajectoire) et jeunes en situation d'insertion durable dans l'informel.

## 4. Importance systémique

- **Vulnérabilité aux chocs externes :** Une économie dont 68% de l'emploi jeune est concentré dans l'agriculture et le commerce informel présente une exposition élevée aux chocs des matières premières et aux chocs climatiques. La faible diversification est un facteur documenté de volatilité macroéconomique.
- **Productivité agrégée structurellement faible :** La concentration de la main-d'œuvre jeune dans des secteurs à faible valeur ajoutée formelle maintient la productivité du travail à un niveau inférieur au potentiel, limitant la croissance des salaires réels et l'émergence d'une classe moyenne.
- **Inégalités structurelles :** La dualité formel-informel génère des inégalités de revenus, de protection sociale et de droits du travail corrélées avec le genre, le niveau d'éducation et la géographie.
- **Dividende démographique conditionnel :** La croissance de la population active jeune ne se convertira en dividende démographique que si elle s'accompagne d'une migration sectorielle vers des emplois à productivité croissante. Les données indiquent que cette condition n'est pas remplie.

## 5. Implications politiques

**Court terme (0–2 ans) — priorité haute :**
- Lancer des programmes d'extension agricole ciblant les jeunes agriculteurs (semences améliorées, accès aux intrants, formation en gestion) pour augmenter la productivité et les revenus dans le secteur dominant, sans attendre la transformation structurelle.
- Formaliser progressivement les acteurs du petit commerce via des régimes fiscaux simplifiés (micro-entreprise) et des services d'appui (micro-crédit, formation en gestion), en priorité dans les marchés urbains d'Abidjan et Bouaké.

**Moyen terme (2–5 ans) — priorité haute :**
- Développer des zones économiques spéciales ou corridors agro-industriels (transformation du cacao, cajou, caoutchouc, hévéa) pour créer de l'emploi formel à proximité des zones de production, réduisant simultanément les coûts logistiques et le chômage régional.
- Définir et publier une stratégie nationale de diversification sectorielle avec des objectifs quantifiés de création d'emploi formel dans le manufacturing et les services modernes à horizon 2030–2035, assortie d'indicateurs de suivi annuels.

**Moyen terme (2–5 ans) — priorité moyenne :**
- Investir dans les infrastructures logistiques critiques (routes rurales, capacités de stockage à froid, fiabilité de l'énergie) pour réduire les coûts de transformation et renforcer la compétitivité des secteurs intermédiaires.

**Hiérarchie par impact estimé :** agro-industrie (transformation structurelle) > formalisation commerciale > extension agricole > infrastructure logistique.

# INSIGHT 4 — Le Mismatch Index à 47.9% : Une Mesure Composite de l'Inadéquation Structurelle

## 1. Constats observés

- Le mismatch index construit dans le notebook 05 (différence entre part de l'emploi informel et part de l'emploi formel) atteint **47.9 points de pourcentage** (informel : 53.5% ; formel : 5.6% ; mixte : ~40.9%).
- Ce score dépasse le seuil critique défini dans le notebook (> 0.4), déclenchant la qualification : "Marché formel trop étroit pour absorber les jeunes éduqués".
- Le secteur mixte (commerce, manufacturing, transport) représente environ 40% de l'emploi jeune — une part substantielle dont le statut formel/informel effectif reste indéterminé.

## 2. Interprétation structurelle

**Hypothèse 1 — Le mismatch comme résultante de deux blocages indépendants :**  
Le score de 47.9 points peut résulter de deux dynamiques distinctes qu'il est analytiquement important de séparer : (a) l'insuffisance de création d'emploi formel par le secteur privé, liée aux barrières à l'investissement et à la faible diversification économique ; (b) la sur-représentation de l'informel due à l'absence de politiques de formalisation efficaces et à des coûts de conformité trop élevés pour les TPE/PME. Ces deux mécanismes appellent des politiques différentes et ne peuvent être traités par un seul instrument.

**Hypothèse 2 — Le secteur mixte comme zone d'intervention prioritaire :**  
Les 40% d'emploi classé "mixte" (commerce, manufacturing, transport) représentent potentiellement la zone de transition la plus accessible vers le formel. Ces secteurs combinent une présence significative et une hétérogénéité structurelle permettant des politiques de formalisation graduelles. Le manufacturing (8.1%) est particulièrement stratégique comme vecteur de création d'emploi formel à moyen terme.

## 3. Limites analytiques

- **Sensibilité du score à la classification :** Le mismatch index de 47.9% est entièrement dépendant de la liste de mots-clés définissant "formel" et "informel". Une classification alternative incluant le manufacturing dans le formel (argument défendable pour les grandes entreprises manufacturières) réduirait significativement le score. L'indicateur doit être interprété comme un ordre de grandeur et une direction, non comme une mesure précise.
- **Absence de benchmark international :** Sans comparaison avec des pays à niveau de développement similaire (Ghana, Sénégal, Kenya), il est impossible de qualifier si 47.9 points est un niveau exceptionnellement élevé ou simplement représentatif d'une économie africaine en développement.
- **Non-prise en compte de la qualité de l'emploi formel :** L'emploi public (administration, éducation) — principal composant du "formel" dans les données — présente des dynamiques de recrutement très différentes de l'emploi privé formel. La distinction est analytiquement importante pour les politiques d'emploi.

## 4. Importance systémique

- **Signal de cohérence analytique :** Le mismatch index confirme de manière indépendante le paradoxe du diplômé (Insight 1) : si seulement 5.6% de l'emploi est formel, il est mathématiquement impossible que le marché du travail absorbe productivement des taux croissants de diplômés du supérieur, quelle que soit la qualité des formations.
- **Cadrage pour les projections 2035 :** Si le taux de formalisation de l'emploi ne s'améliore pas significativement d'ici 2035, la pression démographique aggravera mécaniquement le mismatch : davantage de jeunes diplômés pour un secteur formel structurellement stagnant.

## 5. Implications politiques

**Court terme (0–2 ans) :**
- Valider et affiner le mismatch index en le confrontant à des données administratives (registres d'entreprises, données CNPS de couverture sociale) pour améliorer la fiabilité de la classification formel/informel.
- Publier le mismatch index comme indicateur de pilotage officiel dans les rapports annuels du Ministère de l'Emploi, avec un objectif de réduction à 10 ans.

**Moyen terme (2–5 ans) :**
- Définir une trajectoire de formalisation sectorielle prioritaire, en ciblant d'abord le secteur "mixte" (manufacturing, transport, commerce structuré) comme zone de transition la plus accessible, avant le secteur agricole qui requiert des transformations structurelles plus profondes.
- Conditionner l'accès aux marchés publics et aux financements de développement à des critères de formalité (enregistrement, couverture sociale minimale des employés).

# INSIGHT 5 — La Tendance 2015–2025 : Stabilité Structurelle et Découplage Croissance–Emploi

## 1. Constats observés

- Sur la période 2015–2025, le taux d'inactivité moyen se maintient dans une fourchette haute (~0.38–0.46), avec une décroissance partielle depuis 2018–2019, sans rupture de tendance significative.
- Le taux de chômage moyen reste structurellement bas (~0.08–0.11), stable sur la décennie, avec une légère tendance à la hausse sur 2022–2025.
- Aucune convergence des deux indicateurs n'est observable : l'inactivité reste dominante et le chômage officiel marginal, reflétant la structure de l'économie informelle.
- La vulnérabilité moyenne sur la période entière (2015–2025) par niveau d'éducation confirme la hiérarchie paradoxale observée en coupe 2025, suggérant que ce paradoxe n'est pas récent mais structurellement installé.

## 2. Interprétation structurelle

**Hypothèse 1 — Taux de chômage bas comme artefact de la définition BIT dans un contexte d'absence de protection sociale :**  
Dans une économie sans allocation chômage significative, l'option "être au chômage" (disponible ET en recherche active) est une stratégie que peu peuvent se permettre sur la durée. La quasi-totalité des individus sans emploi productif bascule dans l'inactivité déclarée ou le sous-emploi informel de survie. Le taux de chômage officiel mesure donc davantage la capacité à attendre (dépendant de ressources familiales) que l'ampleur réelle de l'inadéquation emploi-demande.

**Hypothèse 2 — Découplage croissance–emploi formel :**  
L'absence de rupture de tendance sur 10 ans — malgré une croissance du PIB ivoirien positive et soutenue sur la période (parmi les plus élevées d'Afrique subsaharienne) — confirme que la croissance économique récente n'a pas produit les effets d'entraînement attendus sur l'emploi formel jeune. Ce découplage est cohérent avec une croissance tirée par des secteurs capital-intensifs ou par des activités à faible création d'emploi net (extractif, services financiers, BTP grands travaux à main-d'œuvre étrangère).

## 3. Limites analytiques

- **Données moyennes agrégées :** L'analyse temporelle porte sur des moyennes nationales (genre × éducation), masquant des dynamiques potentiellement divergentes par région, tranche d'âge ou secteur.
- **Absence de variables macroéconomiques de contrôle :** La tendance emploi n'est pas mise en relation avec le cycle économique, les chocs (COVID-19 2020–2021, instabilité régionale), les réformes politiques ou les investissements publics sur la période. L'interprétation causale reste impossible.
- **Cohérence méthodologique longitudinale non vérifiable :** Les changements de méthodologie des enquêtes BIT sur 10 ans peuvent introduire des discontinuités artificielles dans les séries.

## 4. Importance systémique

- **Signal d'alerte pour les projections 2035 :** Si la tendance 2015–2025 se prolonge sans rupture de politique publique, les taux d'inactivité resteront à des niveaux élevés, avec un impact cumulatif sur le dividende démographique. La pression démographique croissante aggravera mécaniquement les indicateurs de vulnérabilité en valeur absolue, même si les taux relatifs se stabilisent.
- **Réfutation empirique de l'hypothèse "croissance d'abord" :** La stabilité des indicateurs d'emploi malgré la croissance économique réfute l'hypothèse implicite selon laquelle la croissance macroéconomique suffit à résoudre le problème de l'emploi jeune sans politiques actives ciblées.

## 5. Implications politiques

**Court terme (0–2 ans) :**
- Mettre en place un tableau de bord trimestriel public de suivi des indicateurs d'emploi jeune désagrégés (genre × éducation × secteur × tranche d'âge), comme outil de pilotage et de redevabilité.
- Évaluer rigoureusement les programmes d'emploi jeune existants (AGEFOP, FDFP, FNS) avec des méthodes d'évaluation d'impact robustes (randomisation ou quasi-expérimental), afin de distinguer les programmes efficaces des programmes inefficaces.

**Moyen terme (2–5 ans) :**
- Conditionner les stratégies de croissance sectorielle à des critères d'intensité de création d'emploi formel jeune, rompant explicitement avec le modèle de croissance sans emploi observé sur 2015–2025.
- Intégrer les indicateurs de vulnérabilité composites (chômage + inactivité) et le mismatch index comme métriques officielles complémentaires dans les publications statistiques nationales (INS), afin de donner une image plus complète de l'exclusion du marché du travail.
